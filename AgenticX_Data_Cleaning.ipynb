## Data Quality Issues Identified

During the initial data quality audit, the following issues were identified:

1. **Missing values**
   - Several columns contain missing values, including CustomerID, Name, Gender, Age, City, Signup_Date, Last_purchase_date, purchase_amount, feedback_score, email, Phone_number, and Country.

2. **Inconsistent Gender values**
   - Gender is represented using multiple formats such as `M`, `Male`, `male`, `F`, `Female`, and `female`.

3. **Invalid Age values**
   - The Age column contains invalid values such as `-5` and `200`, which fall outside a reasonable human-age range.

4. **Invalid Feedback Scores**
   - The feedback_score column contains values such as `-1` and `10`.
   - The observed valid rating range is 1–5.

5. **Placeholder values in purchase_amount**
   - The value `-999` occurs frequently and is treated as a placeholder/missing value rather than a genuine purchase amount.

6. **Invalid date values**
   - Signup_Date contains values such as `not_a_date`.
   - Last_purchase_date contains values such as `missing`.
   - The date columns are currently stored as strings.

7. **Inconsistent CustomerID formatting**
   - CustomerIDs appear in mixed formats, such as `C2`, `C4`, `3`, and `5`.
   - No duplicate non-missing CustomerIDs were identified.

8. **Invalid email formats**
   - Some email values are missing the `@` symbol or otherwise do not match the expected email structure.

9. **Invalid phone numbers**
   - Some Phone_number values contain invalid entries such as `abc123` and `0`.
   - Valid-looking entries contain 10 digits.

10. **Duplicate records**
    - No completely duplicated rows were found in the dataset.

```python
# Standardize Gender values

df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "male": "Male",
    "F": "Female",
    "female": "Female"
})

print("Gender values after standardization:")
print(df["Gender"].value_counts(dropna=False))
```

```python
# Replace invalid age values with missing values

df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = np.nan

print("Invalid age values replaced with NaN.")
print("\nAge summary after cleaning:")
print(df["Age"].describe())
```

```python
# Replace invalid feedback scores with missing values

df.loc[
    (df["feedback_score"] < 1) | (df["feedback_score"] > 5),
    "feedback_score"
] = np.nan

print("Invalid feedback scores replaced with NaN.")

print("\nFeedback score values after cleaning:")
print(df["feedback_score"].value_counts(dropna=False).sort_index())
```

```python
# Replace the placeholder purchase amount with a missing value

df["purchase_amount"] = df["purchase_amount"].replace(-999, np.nan)

print("Placeholder purchase amounts replaced with NaN.")

print("\nPurchase amount summary after cleaning:")
print(df["purchase_amount"].describe())
```

```python
# Convert Signup_Date to datetime

df["Signup_Date"] = pd.to_datetime(
    df["Signup_Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Signup_Date data type:")
print(df["Signup_Date"].dtype)

print("\nSample Signup_Date values:")
print(df["Signup_Date"].head(10))
```

```python
# Convert Last_purchase_date to datetime

df["Last_purchase_date"] = pd.to_datetime(
    df["Last_purchase_date"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Last_purchase_date data type:")
print(df["Last_purchase_date"].dtype)

print("\nSample Last_purchase_date values:")
print(df["Last_purchase_date"].head(10))
```

```python
# Identify and replace invalid email addresses

email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

invalid_email_mask = (
    df["email"].notna() &
    ~df["email"].str.match(email_pattern)
)

df.loc[invalid_email_mask, "email"] = np.nan

print("Invalid email addresses replaced with NaN.")

print("\nNumber of missing emails after cleaning:")
print(df["email"].isna().sum())
```

```python
# Identify and replace invalid phone numbers

phone_pattern = r"^\d{10}$"

invalid_phone_mask = (
    df["Phone_number"].notna() &
    ~df["Phone_number"].str.match(phone_pattern)
)

df.loc[invalid_phone_mask, "Phone_number"] = np.nan

print("Invalid phone numbers replaced with NaN.")

print("\nNumber of missing phone numbers after cleaning:")
print(df["Phone_number"].isna().sum())
```

```python
# Check missing values after cleaning invalid data

missing_after_cleaning = df.isnull().sum()

print("Missing values after cleaning:")
print(missing_after_cleaning)
```

```python
# Handle missing categorical values

categorical_missing_columns = ["Gender", "City", "Country"]

for column in categorical_missing_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing categorical values handled.")

print("\nMissing values after categorical cleaning:")
print(df[categorical_missing_columns].isnull().sum())
```

```python
# Calculate median values for numeric columns

age_median = df["Age"].median()
purchase_median = df["purchase_amount"].median()
feedback_median = df["feedback_score"].median()

print("Age median:", age_median)
print("Purchase amount median:", purchase_median)
print("Feedback score median:", feedback_median)
```

```python
# Fill missing numeric values using median

df["Age"] = df["Age"].fillna(age_median)
df["purchase_amount"] = df["purchase_amount"].fillna(purchase_median)
df["feedback_score"] = df["feedback_score"].fillna(feedback_median)

print("Missing numeric values filled using median.")

print("\nMissing values after numeric imputation:")
print(df[["Age", "purchase_amount", "feedback_score"]].isnull().sum())
```

```python
# Check remaining missing values

remaining_missing = df.isnull().sum()

print("Remaining missing values:")
print(remaining_missing)
```

```python
# Handle missing customer names

df["Name"] = df["Name"].fillna("Unknown")

print("Missing customer names handled.")

print("\nMissing values in Name:")
print(df["Name"].isnull().sum())
```

```python
# Check for completely duplicated rows after cleaning

duplicate_rows = df.duplicated().sum()

print("Number of completely duplicated rows:")
print(duplicate_rows)
```

```python
# Inspect completely duplicated rows

duplicate_mask = df.duplicated(keep=False)

print("Total rows involved in duplicates:")
print(duplicate_mask.sum())

print("\nDuplicate rows:")
display(df[duplicate_mask].sort_values(
    by=["Name", "Age", "City"]
))
```

```python
# Remove completely duplicated rows

before_duplicates = len(df)

df = df.drop_duplicates()

after_duplicates = len(df)

print("Rows before removing duplicates:", before_duplicates)
print("Rows after removing duplicates:", after_duplicates)
print("Duplicate rows removed:", before_duplicates - after_duplicates)
```

```python
# Verify that no completely duplicated rows remain

print("Remaining duplicate rows:")
print(df.duplicated().sum())
```

```python
# Detect potential outliers in purchase_amount using IQR

Q1 = df["purchase_amount"].quantile(0.25)
Q3 = df["purchase_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["purchase_amount"] < lower_bound) |
    (df["purchase_amount"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of potential outliers:", len(outliers))
```

```python
df["purchase_amount"] = df["purchase_amount"].fillna(2460.355)
```

```python
# Inspect the distribution of purchase_amount after cleaning

print("Purchase amount statistics:")
print(df["purchase_amount"].describe())

print("\nNumber of values equal to the median:")
print((df["purchase_amount"] == purchase_median).sum())
```

```python
# Select original purchase amounts for outlier analysis
# Exclude values that were created through median imputation

original_purchase = df.loc[
    df["purchase_amount"] != purchase_median,
    "purchase_amount"
]

print("Number of original purchase values:")
print(len(original_purchase))

print("\nOriginal purchase amount statistics:")
print(original_purchase.describe())
```

```python
# Calculate IQR boundaries using original purchase values

Q1_purchase = original_purchase.quantile(0.25)
Q3_purchase = original_purchase.quantile(0.75)

IQR_purchase = Q3_purchase - Q1_purchase

lower_bound_purchase = Q1_purchase - 1.5 * IQR_purchase
upper_bound_purchase = Q3_purchase + 1.5 * IQR_purchase

print("Q1:", Q1_purchase)
print("Q3:", Q3_purchase)
print("IQR:", IQR_purchase)
print("Lower bound:", lower_bound_purchase)
print("Upper bound:", upper_bound_purchase)

potential_outliers = original_purchase[
    (original_purchase < lower_bound_purchase) |
    (original_purchase > upper_bound_purchase)
]

print("\nNumber of potential outliers:", len(potential_outliers))
```

## Outlier Analysis

Outlier detection was performed on the original observed purchase amounts before median imputation.

The Interquartile Range (IQR) method was used:

- Q1 = 1159.385
- Q3 = 3733.880
- IQR = 2574.495
- Lower bound = -2702.3575
- Upper bound = 7595.6225

The observed purchase amounts range from 10.31 to 4998.67. Since no value falls outside the calculated IQR boundaries, no potential outliers were detected.

Therefore, no purchase amount values were removed as outliers.

```python
# Load the original dataset again for Age outlier analysis

raw_df = pd.read_csv("messy_customer_data.csv")

original_age = raw_df["Age"].dropna()

print("Number of original Age values:", len(original_age))
print("\nOriginal Age statistics:")
print(original_age.describe())
```

```python
# Keep only valid human ages for outlier analysis

valid_age = original_age[
    (original_age >= 18) &
    (original_age <= 100)
]

print("Number of valid age values:", len(valid_age))

print("\nValid age statistics:")
print(valid_age.describe())
```

```python
# Calculate IQR boundaries for valid ages

Q1_age = valid_age.quantile(0.25)
Q3_age = valid_age.quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_bound_age = Q1_age - 1.5 * IQR_age
upper_bound_age = Q3_age + 1.5 * IQR_age

print("Q1:", Q1_age)
print("Q3:", Q3_age)
print("IQR:", IQR_age)
print("Lower bound:", lower_bound_age)
print("Upper bound:", upper_bound_age)

age_outliers = valid_age[
    (valid_age < lower_bound_age) |
    (valid_age > upper_bound_age)
]

print("\nNumber of potential age outliers:", len(age_outliers))
```

### Age Outlier Analysis

For outlier analysis, only valid ages between 18 and 100 were considered. The clearly invalid values `-5` and `200` were excluded because they had already been identified as invalid data values.

Using the IQR method:

- Q1 = 31
- Q3 = 58
- IQR = 27
- Lower bound = -9.5
- Upper bound = 98.5

The valid ages range from 18 to 70, so no valid age falls outside the calculated IQR boundaries.

Therefore, no additional statistical outliers were detected in the valid Age values.

```python
# Load original feedback scores for outlier analysis

original_feedback = raw_df["feedback_score"].dropna()

print("Number of original feedback scores:", len(original_feedback))

print("\nOriginal feedback statistics:")
print(original_feedback.describe())
```

```python
# Calculate IQR boundaries for valid feedback scores

Q1_feedback = valid_feedback.quantile(0.25)
Q3_feedback = valid_feedback.quantile(0.75)

IQR_feedback = Q3_feedback - Q1_feedback

lower_bound_feedback = Q1_feedback - 1.5 * IQR_feedback
upper_bound_feedback = Q3_feedback + 1.5 * IQR_feedback

print("Q1:", Q1_feedback)
print("Q3:", Q3_feedback)
print("IQR:", IQR_feedback)
print("Lower bound:", lower_bound_feedback)
print("Upper bound:", upper_bound_feedback)

feedback_outliers = valid_feedback[
    (valid_feedback < lower_bound_feedback) |
    (valid_feedback > upper_bound_feedback)
]

print("\nNumber of potential feedback outliers:", len(feedback_outliers))
```
